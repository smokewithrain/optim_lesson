```python
import numpy as np
import random
```

```python
def dp(capacity, item_weights, item_values):
    """
    动态规划求解01背包问题
    
    :param capacity: 背包容量
    :param item_weights: 物品重量列表
    :param item_values: 物品价值列表
    :return value: 最优价值
    :return selected: 最优价值下被选中的物品索引
    """
    n = len(item_weights)
    # dp[i][j]表示前i个物品在承重j下的最大价值
    # 即分解子问题为对前i个问题，背包容量j时，求得最优解的价值
    # dp存储子问题的答案，后一步在前一步基础上计算，推到最后求出对所有物品，背包容量capacity时的最优解
    # 显然，重量和价值都是离散的整数值
    dp = np.zeros((n+1, capacity+1), dtype=int)  # 初始化为0
    for i in range(1, n+1):                      # 逐个考虑物体
        for j in range(1, capacity+1):           # 逐个考虑容量
            if item_weights[i-1] > j:            # 若当前物品重量大于背包承重
                dp[i][j] = dp[i-1][j]            # 不放入
            else:                                # 若当前物品重量小于等于背包承重
                                                 # 取放物品和不放物品的最大价值，作为dp[i][j]
                dp[i][j] = max(dp[i-1][j], dp[i-1][j-item_weights[i-1]] + item_values[i-1])
    value = dp[n][capacity] 
    # 找到最优解下选中的物品
    remain_capacity = capacity  
    selected = []               # 选中的物品索引
    for i in range(n, 0, -1):   # 从n到1物品，根据dp结果选择装入背包，当背包被装满，则退出
        if dp[i][remain_capacity] != dp[i-1][remain_capacity]:  # 该物品被选中
            selected.append(i-1)
            remain_capacity -= item_weights[i-1]                # 背包剩余空间
            if remain_capacity == 0:
                break 
    selected = selected[::-1]
    return value, selected
```

```python
capacity = 100          # 背包容量
num_items = 15          # 物品数量
min_weight = 1          # 物品最小重量
max_weight = 15         # 物品最大重量
min_value = 10          # 物品最小价值
max_value = 100         # 物品最大价值

# 随机生成物品重量、价值, 且背包无法全部装下
while True:
    item_weights = [random.randint(min_weight, max_weight) for _ in range(num_items)]   
    item_values = [random.randint(min_value, max_value) for _ in range(num_items)]   

    if sum(item_weights) > capacity:
        break
print(f"背包容量: {capacity}")
print(f"物品重量列表: {item_weights}")
print(f"物品总重量: {sum(item_weights)}")
print(f"物品价值列表: {item_values}")
print(f"物品总价值: {sum(item_values)}")

value, selected = dp(capacity, item_weights, item_values)
print("\n动态规划求解01背包问题")
print(f"选中的物品索引: {selected}")
print(f"选中的物品总重量: {sum([item_weights[i] for i in selected])}")
print(f"最优价值: {value}")
```
